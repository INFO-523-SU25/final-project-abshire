```python
## COMPLETE CREDIT TO: https://github.com/stephen1694/pybaseball/blob/main/download_2015_2024.py

from pybaseball import statcast
import pandas as pd
from calendar import monthrange

def download_statcast_data(year, start_day, end_day, batter_ids=None):
    """
    Download Statcast data for a given year, with optional batter filter applied immediately.

    Parameters:
    - year: String, last two digits of the year (e.g., '24' for 2024)
    - start_day: String, MM-DD (e.g., '03-01')
    - end_day: String, MM-DD (e.g., '09-30')
    - batter_ids: List of batter IDs to keep (optional). If None, keeps all.
    """
    start_date = f"20{year}-{start_day}"
    end_date = f"20{year}-{end_day}"
    start_month = int(start_day.split('-')[0])
    end_month = int(end_day.split('-')[0])
    list_months = [f"{m:02d}" for m in range(start_month, end_month + 1)]
    
    list_dfs = []
    for i, month in enumerate(list_months):
        if i == 0:
            start = start_date
            _, last_day = monthrange(int(f"20{year}"), int(month))
            end = f"20{year}-{month}-{last_day:02d}" if len(list_months) > 1 else end_date
        elif i == len(list_months) - 1:
            start = f"20{year}-{month}-01"
            end = end_date
        else:
            start = f"20{year}-{month}-01"
            _, last_day = monthrange(int(f"20{year}"), int(month))
            end = f"20{year}-{month}-{last_day:02d}"

        print(f"Downloading data from {start} to {end}")
        data = statcast(start_dt=start, end_dt=end)

        # Filter early to save memory
        if batter_ids is not None:
            data = data[data['batter'].isin(batter_ids)]

        list_dfs.append(data)

    finalDf = pd.concat(list_dfs).drop_duplicates()
    return finalDf

# Your full Top 100 batter IDs
top_100_batters = [
    665833, 592450, 656941, 660271, 665742, 646240, 695578, 621566, 624413, 592518,
    641857, 650490, 677951, 665487, 700250, 666018, 608369, 680718, 656305, 683002,
    701762, 665489, 680776, 682998, 666969, 663586, 621439, 683734, 691023, 673548,
    664040, 660821, 663457, 663728, 691406, 671218, 607043, 668227, 502054, 670242,
    686948, 682663, 547180, 681481, 681351, 665966, 571970, 677594, 691016, 596115,
    672386, 518692, 683737, 677800, 672580, 606466, 669257, 666134, 671056, 663886,
    663616, 666176, 669065, 545361, 669016, 666182, 682829, 676475, 668804, 621043,
    687263, 694671, 668901, 667670, 696100, 700337, 572233, 657757, 672820, 596146,
    608324, 661388, 668930, 553993, 542303, 691176, 671732, 575929, 676572, 656811,
    663757, 667472, 521692, 673237, 650402, 621493, 656555, 663656, 592885, 593871,
    677950, 676059, 686469
    
]

# Example usage for 2024 with batter filter
pitch_2024_df = download_statcast_data('24', '03-01', '09-30', batter_ids=top_100_batters)

# Drop Seoul series and other spring games
pitch_2024_df = pitch_2024_df[
    ~((pitch_2024_df['game_date'] == '2024-03-20') & (~pitch_2024_df['home_team'].isin(['LAD', 'SD'])))
]
pitch_2024_df = pitch_2024_df[
    ~((pitch_2024_df['game_date'] == '2024-03-21') & (~pitch_2024_df['home_team'].isin(['LAD', 'SD'])))
]
dates_to_drop_24 = ['2024-03-22','2024-03-23','2024-03-24','2024-03-25','2024-03-26','2024-03-27']
pitch_2024_df = pitch_2024_df[~pitch_2024_df['game_date'].isin(dates_to_drop_24)]

pitch_2024_df.to_csv('pitch_2024_top100.csv', index=False)
```

### Merge batter_id from razzball resource and make pitch_group

```python
import pandas as pd

# Load data
pitch_df = pd.read_csv('/workspaces/final-project-abshire/notebooks/pitch_2024_top100_reduced.csv')
player_ids_df = pd.read_csv('/workspaces/final-project-abshire/data/player_ids.csv')

# Prepare player IDs
player_ids_sub = player_ids_df[['MLBAMID', 'Name']].rename(columns={'Name': 'batter_name'})

# Merge on batter and MLBAMID
df_top100 = pitch_df.merge(player_ids_sub, left_on='batter', right_on='MLBAMID', how='left')
df_top100 = df_top100.drop(columns=['MLBAMID'])
df_top100 = df_top100.rename(columns={'player_name': 'pitcher_name'})

# Filter to hitting events only
hitting_events = ['single', 'double', 'triple', 'home_run']
df_hits = df_top100[df_top100['events'].isin(hitting_events)].copy()

# Now define your pitch_type_map
pitch_type_map = {
    'FF': 'Fastball',  # Four-Seam
    'FA': 'Fastball',  # General Fastball
    'FT': 'Fastball',  # Two-Seam
    'SI': 'Fastball',  # Sinker
    'FS': 'Fastball',  # Split Finger
    'SL': 'Slider',    # Slider
    'SV': 'Slider',    # Slurve
    'ST': 'Curveball', # Sweeper
    'CU': 'Curveball', # Curveball
    'KC': 'Curveball', # Knuckle Curve
    'CH': 'Changeup',  # Change Up
    'FO': 'Changeup',  # Fork Ball
    'EP': 'Changeup',  # Eephus
    'FC': 'Fastball',  # Cutter
    'KN': 'Knuckleball', # Knuckleball
    'SC': 'Other',     # Screwball
    'CS': 'Other',     # Slow Curve
    'PO': 'Other',     # Pitch Out?
    None: 'Unknown',   # or np.nan mapped to 'Unknown'
    'NaN': 'Unknown'   # if string 'NaN'
}

# Map pitch_type to pitch_group column
df_top100['pitch_group'] = df_top100['pitch_type'].map(pitch_type_map).fillna('Unknown')

print(df_top100.head())

# Unique count of batter_name
unique_batter_count = df_top100['batter_name'].nunique()
print(f"Unique batter_name count: {unique_batter_count}")

# Print the unique batter names
unique_batters = df_top100['batter_name'].unique()
print("Unique batter names:")
print(unique_batters)
```
